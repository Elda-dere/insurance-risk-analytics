```python
import pandas as pd

# Load with pipe delimiter
df = pd.read_csv('data/MachineLearningRating_v3.txt', delimiter='|')

print(f"✅ Data loaded successfully!")
print(f"Shape: {df.shape} rows, {df.shape[1]} columns")
print(f"\nColumn names (first 10):")
print(df.columns.tolist()[:10])
df.head()
```

```python
# Convert TotalClaims to numeric (they might be strings with dots)
df['TotalClaims'] = pd.to_numeric(df['TotalClaims'], errors='coerce')
df['TotalPremium'] = pd.to_numeric(df['TotalPremium'], errors='coerce')

# Calculate Loss Ratio
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
df['Margin'] = df['TotalPremium'] - df['TotalClaims']
df['HasClaim'] = (df['TotalClaims'] > 0).astype(int)

print(f"Overall Loss Ratio: {df['LossRatio'].mean():.4f}")
print(f"Overall Margin: R{df['Margin'].mean():.2f}")
print(f"Claim Frequency: {df['HasClaim'].mean():.2%}")
```

```python
# Convert TotalClaims to numeric (they might be strings with dots)
df['TotalClaims'] = pd.to_numeric(df['TotalClaims'], errors='coerce')
df['TotalPremium'] = pd.to_numeric(df['TotalPremium'], errors='coerce')

# Calculate Loss Ratio
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
df['Margin'] = df['TotalPremium'] - df['TotalClaims']
df['HasClaim'] = (df['TotalClaims'] > 0).astype(int)

print(f"Overall Loss Ratio: {df['LossRatio'].mean():.4f}")
print(f"Overall Margin: R{df['Margin'].mean():.2f}")
print(f"Claim Frequency: {df['HasClaim'].mean():.2%}")
```

```python
# Calculate loss ratio by province
loss_by_province = df.groupby('Province').agg({
    'TotalClaims': 'sum',
    'TotalPremium': 'sum'
})
loss_by_province['LossRatio'] = loss_by_province['TotalClaims'] / loss_by_province['TotalPremium']
loss_by_province = loss_by_province.sort_values('LossRatio', ascending=False)

print("Loss Ratio by Province:")
print(loss_by_province)
```

```python
import pandas as pd
import numpy as np

# Load data with pipe delimiter
df = pd.read_csv('data/MachineLearningRating_v3.txt', delimiter='|')

# Fix numeric columns (convert problematic strings)
numeric_cols = ['TotalPremium', 'TotalClaims', 'CalculatedPremiumPerTerm', 
                'CustomValueEstimate', 'CapitalOutstanding']

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# Calculate key metrics
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium'].replace(0, np.nan)
df['Margin'] = df['TotalPremium'] - df['TotalClaims']
df['HasClaim'] = (df['TotalClaims'] > 0).astype(int)

print(f"✅ Data loaded! Shape: {df.shape}")
print(f"Columns: {len(df.columns)}")
print(f"\nOverall Loss Ratio: {df['LossRatio'].mean():.4f}")
print(f"Overall Margin: R{df['Margin'].mean():.2f}")
```
